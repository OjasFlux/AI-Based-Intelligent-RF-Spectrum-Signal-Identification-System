```python
# ============================================================
# AI Based Intelligent RF Spectrum Signal Identification System
# Notebook: 01_dataset_analysis.ipynb
# Purpose: RadioML 2016.10a Dataset Analysis
# ============================================================

# -----------------------------
# 1. IMPORT LIBRARIES
# -----------------------------

import os
import pickle
import numpy as np
import matplotlib.pyplot as plt

print("=" * 60)
print("LIBRARIES IMPORTED SUCCESSFULLY")
print("=" * 60)


# -----------------------------
# 2. DATASET PATH
# -----------------------------

dataset_path = "../data/dataset/RML2016.10a_dict.pkl"

print("\nDataset Path:")
print(dataset_path)

# Check whether dataset exists
if not os.path.exists(dataset_path):
    print("\nERROR: Dataset file not found!")
    print("Please check the dataset location and filename.")
    print("\nExpected location:")
    print("../data/raw/RML2016.10a_dict.pkl")
else:
    print("\nDataset file found successfully!")


# -----------------------------
# 3. LOAD DATASET
# -----------------------------

if os.path.exists(dataset_path):

    print("\nLoading RadioML 2016.10a dataset...")

    with open(dataset_path, "rb") as file:
        dataset = pickle.load(file, encoding="latin1")

    print("Dataset loaded successfully!")


# -----------------------------
# 4. BASIC DATASET INFORMATION
# -----------------------------

if os.path.exists(dataset_path):

    print("\n" + "=" * 60)
    print("BASIC DATASET INFORMATION")
    print("=" * 60)

    print("Dataset Type:", type(dataset))
    print("Number of Dataset Entries:", len(dataset))


# -----------------------------
# 5. EXTRACT MODULATION TYPES
# -----------------------------

if os.path.exists(dataset_path):

    modulation_types = sorted(list(set(key[0] for key in dataset.keys())))

    print("\n" + "=" * 60)
    print("MODULATION TYPES")
    print("=" * 60)

    print("Number of Modulation Classes:", len(modulation_types))

    for i, modulation in enumerate(modulation_types, start=1):
        print(f"{i}. {modulation}")


# -----------------------------
# 6. EXTRACT SNR VALUES
# -----------------------------

if os.path.exists(dataset_path):

    snr_values = sorted(list(set(key[1] for key in dataset.keys())))

    print("\n" + "=" * 60)
    print("SNR VALUES")
    print("=" * 60)

    print("Number of SNR Levels:", len(snr_values))
    print("SNR Values:", snr_values)


# -----------------------------
# 7. DISPLAY DATASET KEYS
# -----------------------------

if os.path.exists(dataset_path):

    keys = list(dataset.keys())

    print("\n" + "=" * 60)
    print("FIRST 10 DATASET KEYS")
    print("=" * 60)

    for key in keys[:10]:
        print(key)


# -----------------------------
# 8. CHECK IQ SAMPLE SHAPE
# -----------------------------

if os.path.exists(dataset_path):

    sample_key = keys[0]
    sample_data = dataset[sample_key]

    print("\n" + "=" * 60)
    print("IQ SAMPLE INFORMATION")
    print("=" * 60)

    print("Sample Key:", sample_key)
    print("Sample Data Shape:", sample_data.shape)

    print("\nInterpretation:")

    if len(sample_data.shape) == 3:
        print("Number of Signals:", sample_data.shape[0])
        print("IQ Components:", sample_data.shape[1])
        print("Samples per Signal:", sample_data.shape[2])


# -----------------------------
# 9. CALCULATE TOTAL SIGNALS
# -----------------------------

if os.path.exists(dataset_path):

    total_signals = sum(data.shape[0] for data in dataset.values())

    print("\n" + "=" * 60)
    print("TOTAL DATASET STATISTICS")
    print("=" * 60)

    print("Total Dataset Entries:", len(dataset))
    print("Total RF Signal Samples:", total_signals)
    print("Number of Modulation Classes:", len(modulation_types))
    print("Number of SNR Levels:", len(snr_values))


# -----------------------------
# 10. SELECT ONE SIGNAL
# -----------------------------

if os.path.exists(dataset_path):

    selected_modulation = modulation_types[0]
    selected_snr = snr_values[len(snr_values) // 2]

    selected_key = (selected_modulation, selected_snr)

    signal = dataset[selected_key][0]

    I = signal[0]
    Q = signal[1]

    print("\n" + "=" * 60)
    print("SELECTED SIGNAL INFORMATION")
    print("=" * 60)

    print("Modulation Type:", selected_modulation)
    print("SNR:", selected_snr, "dB")
    print("Signal Shape:", signal.shape)
    print("Number of IQ Samples:", len(I))


# -----------------------------
# 11. PLOT I AND Q SIGNALS
# -----------------------------

if os.path.exists(dataset_path):

    sample_numbers = np.arange(len(I))

    plt.figure(figsize=(12, 5))

    plt.plot(sample_numbers, I, label="I - In-phase")
    plt.plot(sample_numbers, Q, label="Q - Quadrature")

    plt.title(
        f"IQ Signal - {selected_modulation} at {selected_snr} dB"
    )

    plt.xlabel("Sample Number")
    plt.ylabel("Amplitude")
    plt.legend()
    plt.grid(True)

    plt.show()


# -----------------------------
# 12. PLOT I COMPONENT
# -----------------------------

if os.path.exists(dataset_path):

    plt.figure(figsize=(12, 4))

    plt.plot(I)

    plt.title(
        f"In-phase (I) Component - {selected_modulation} at {selected_snr} dB"
    )

    plt.xlabel("Sample Number")
    plt.ylabel("Amplitude")
    plt.grid(True)

    plt.show()


# -----------------------------
# 13. PLOT Q COMPONENT
# -----------------------------

if os.path.exists(dataset_path):

    plt.figure(figsize=(12, 4))

    plt.plot(Q)

    plt.title(
        f"Quadrature (Q) Component - {selected_modulation} at {selected_snr} dB"
    )

    plt.xlabel("Sample Number")
    plt.ylabel("Amplitude")
    plt.grid(True)

    plt.show()


# -----------------------------
# 14. PLOT CONSTELLATION DIAGRAM
# -----------------------------

if os.path.exists(dataset_path):

    plt.figure(figsize=(7, 7))

    plt.scatter(I, Q, alpha=0.6)

    plt.title(
        f"Constellation Diagram - {selected_modulation} at {selected_snr} dB"
    )

    plt.xlabel("In-phase (I)")
    plt.ylabel("Quadrature (Q)")

    plt.grid(True)
    plt.axis("equal")

    plt.show()


# -----------------------------
# 15. CREATE COMPLEX SIGNAL
# -----------------------------

if os.path.exists(dataset_path):

    complex_signal = I + 1j * Q

    print("\n" + "=" * 60)
    print("COMPLEX SIGNAL")
    print("=" * 60)

    print("Complex Signal Shape:", complex_signal.shape)


# -----------------------------
# 16. FFT AND FREQUENCY SPECTRUM
# -----------------------------

if os.path.exists(dataset_path):

    fft_result = np.fft.fft(complex_signal)

    fft_shifted = np.fft.fftshift(fft_result)

    frequency = np.fft.fftfreq(len(complex_signal))
    frequency_shifted = np.fft.fftshift(frequency)

    magnitude = np.abs(fft_shifted)

    plt.figure(figsize=(12, 5))

    plt.plot(frequency_shifted, magnitude)

    plt.title(
        f"Frequency Spectrum using FFT - {selected_modulation} at {selected_snr} dB"
    )

    plt.xlabel("Normalized Frequency")
    plt.ylabel("Magnitude")

    plt.grid(True)

    plt.show()


# -----------------------------
# 17. COMPARE DIFFERENT MODULATIONS
# -----------------------------

if os.path.exists(dataset_path):

    print("\n" + "=" * 60)
    print("MODULATION SIGNAL COMPARISON")
    print("=" * 60)

    available_modulations = modulation_types[:4]

    for modulation in available_modulations:

        key = (modulation, selected_snr)

        if key in dataset:

            comparison_signal = dataset[key][0]

            comparison_I = comparison_signal[0]
            comparison_Q = comparison_signal[1]

            plt.figure(figsize=(6, 6))

            plt.scatter(
                comparison_I,
                comparison_Q,
                alpha=0.5
            )

            plt.title(
                f"Constellation - {modulation} at {selected_snr} dB"
            )

            plt.xlabel("In-phase (I)")
            plt.ylabel("Quadrature (Q)")

            plt.grid(True)
            plt.axis("equal")

            plt.show()


# -----------------------------
# 18. SAVE DATASET SUMMARY
# -----------------------------

if os.path.exists(dataset_path):

    summary_path = "../data/dataset_summary.txt"

    with open(summary_path, "w") as summary_file:

        summary_file.write(
            "AI Based Intelligent RF Spectrum Signal Identification System\n"
        )

        summary_file.write(
            "RadioML 2016.10a Dataset Analysis Summary\n"
        )

        summary_file.write("=" * 60 + "\n\n")

        summary_file.write(
            f"Dataset Type: {type(dataset)}\n"
        )

        summary_file.write(
            f"Number of Dataset Entries: {len(dataset)}\n"
        )

        summary_file.write(
            f"Total Signal Samples: {total_signals}\n"
        )

        summary_file.write(
            f"Number of Modulation Classes: {len(modulation_types)}\n"
        )

        summary_file.write(
            f"Modulation Types: {modulation_types}\n"
        )

        summary_file.write(
            f"Number of SNR Levels: {len(snr_values)}\n"
        )

        summary_file.write(
            f"SNR Values: {snr_values}\n"
        )

        summary_file.write(
            f"IQ Sample Shape: {sample_data.shape}\n"
        )

    print("\nDataset summary saved successfully!")
    print("Location:", summary_path)


# -----------------------------
# 19. FINAL PROJECT ANALYSIS SUMMARY
# -----------------------------

if os.path.exists(dataset_path):

    print("\n" + "=" * 60)
    print("DATASET ANALYSIS COMPLETED SUCCESSFULLY")
    print("=" * 60)

    print(f"Dataset: RadioML 2016.10a")
    print(f"Modulation Classes: {len(modulation_types)}")
    print(f"SNR Levels: {len(snr_values)}")
    print(f"Total Signals: {total_signals}")
    print(f"IQ Sample Shape: {sample_data.shape}")

    print("\nCompleted Analysis:")
    print("[✓] Dataset Loaded")
    print("[✓] Modulation Types Extracted")
    print("[✓] SNR Values Extracted")
    print("[✓] IQ Data Shape Analyzed")
    print("[✓] Total Signals Calculated")
    print("[✓] I and Q Signals Visualized")
    print("[✓] Constellation Diagram Generated")
    print("[✓] FFT Spectrum Generated")
    print("[✓] Dataset Summary Saved")

    print("\nNext Step: Data Preprocessing")
```
